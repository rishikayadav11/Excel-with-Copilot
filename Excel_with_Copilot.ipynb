- Click the **dropdown arrow** of each column.
- Select **(Blanks)** to identify missing values.

 ## Filling Missing Numeric Values in **Units Sold** and **Unit Price** Using Excel

To handle missing numeric values in the Units Sold and Unit Price columns, the Mean (Average) Method was used in Excel. This method replaces blank cells with the average value of the respective column.

**Step 1: Find the Average of Units Sold.**

  (Assuming Units Sold is in column H)
  
  **FORMULA:** =AVERAGE(H2:H351)

  This formula calculates the average of all available values while ignoring blank cells.

- Press **Ctrl + Enter** to fill all selected blank cells at once.
